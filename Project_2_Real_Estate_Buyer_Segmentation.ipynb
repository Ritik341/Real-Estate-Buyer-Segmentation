```python
# Aggregate property details customer-wise
property_summary = (
    properties_clean
    .dropna(subset=['client_ref'])
    .groupby('client_ref')
    .agg(
        properties_count=('client_ref', 'size'),
        avg_sale_price=('sale_price', 'mean'),
        total_sale_value=('sale_price', 'sum'),
        max_sale_price=('sale_price', 'max')
    )
    .reset_index()
)

print("Property aggregation completed!")
print("Total unique buyers with property records:", len(property_summary))

display(property_summary.head(10))
```

```python
# Merge client data with property summary
buyer_data = pd.merge(
    clients_clean,
    property_summary,
    left_on='client_id',
    right_on='client_ref',
    how='left'
)

# Fill missing property records with 0
buyer_data['properties_count'] = buyer_data['properties_count'].fillna(0)

# Check merged dataset
print("Buyer data shape:", buyer_data.shape)
print("Unique clients:", buyer_data['client_id'].nunique())
print("Duplicate client IDs:", buyer_data['client_id'].duplicated().sum())

display(buyer_data.head(10))
```

```python
# Check missing values in each column
missing_values = buyer_data.isnull().sum()

missing_summary = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': (missing_values / len(buyer_data) * 100).round(2)
})

missing_summary = missing_summary[missing_summary['Missing Values'] > 0]

print("Missing Values Analysis:")
display(missing_summary)
```

```python
# Analyze available age data
print("Total clients:", len(buyer_data))
print("Available age records:", buyer_data['age'].notna().sum())
print("Missing age records:", buyer_data['age'].isna().sum())

print("\nAge Statistics:")
display(buyer_data['age'].describe())

print("\nAge Distribution:")
display(buyer_data['age'].value_counts(bins=5).sort_index())
```

```python
# Check missing age by client type
age_missing_analysis = (
    buyer_data
    .groupby('client_type')
    .agg(
        total_clients=('client_id', 'count'),
        available_age=('age', 'count'),
        missing_age=('age', lambda x: x.isna().sum())
    )
)

age_missing_analysis['missing_percentage'] = (
    age_missing_analysis['missing_age'] /
    age_missing_analysis['total_clients'] * 100
).round(2)

display(age_missing_analysis)
```

```python
# Check missing values excluding age and date_of_birth
other_missing = buyer_data.drop(
    columns=['age', 'date_of_birth']
).isnull().sum()

other_missing = other_missing[other_missing > 0]

print("Missing values in other columns:")
display(other_missing)
```

```python
# Identify categorical columns
categorical_cols = buyer_data.select_dtypes(
    include=['object', 'string']
).columns.tolist()

print("Categorical Columns:", categorical_cols)

# Check unique values
for col in categorical_cols:
    print(f"\n{col}:")
    print(buyer_data[col].value_counts())
```

```python
# Select useful categorical features
categorical_features = [
    'client_type',
    'gender',
    'country',
    'region',
    'acquisition_purpose',
    'loan_applied',
    'referral_channel'
]

# Select useful numerical features
numerical_features = [
    'satisfaction_score',
    'properties_count',
    'avg_sale_price',
    'total_sale_value',
    'max_sale_price'
]

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)

print("\nSelected Feature Data:")
display(buyer_data[categorical_features + numerical_features].head())
```

```python
# Check missing values in selected features
selected_features = categorical_features + numerical_features

print("Missing Values:")
display(buyer_data[selected_features].isnull().sum())

print("\nUnique Categories:")
for col in categorical_features:
    print(f"{col}: {buyer_data[col].nunique()} unique values")
```

```python
# Apply One-Hot Encoding
buyer_encoded = pd.get_dummies(
    buyer_data[categorical_features],
    columns=categorical_features,
    drop_first=True,
    dtype=int
)

print("One-Hot Encoding Completed!")
print("Original categorical columns:", len(categorical_features))
print("Encoded columns:", buyer_encoded.shape[1])

display(buyer_encoded.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Select numerical features
numeric_data = buyer_data[numerical_features].copy()

# Apply Standard Scaling
scaler = StandardScaler()

buyer_scaled = pd.DataFrame(
    scaler.fit_transform(numeric_data),
    columns=numerical_features,
    index=buyer_data.index
)

print("Numerical Feature Scaling Completed!")

display(buyer_scaled.head())

print("\nScaled Feature Statistics:")
display(buyer_scaled.describe().round(2))
```

```python
# Combine encoded categorical and scaled numerical features
final_features = pd.concat(
    [buyer_encoded, buyer_scaled],
    axis=1
)

print("Final ML Dataset Created!")
print("Dataset Shape:", final_features.shape)

print("\nMissing Values:")
print(final_features.isnull().sum().sum())

print("\nDuplicate Rows:", final_features.duplicated().sum())

display(final_features.head())
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Calculate WCSS for different cluster values
wcss = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(final_features)
    wcss.append(kmeans.inertia_)

# Plot Elbow Curve
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), wcss, marker='o')

plt.title('Elbow Method for Optimal Clusters')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Inertia)')
plt.xticks(range(2, 11))
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 9):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(final_features)

    score = silhouette_score(
        final_features,
        labels,
        sample_size=1000,
        random_state=42
    )

    silhouette_scores.append(score)

    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
# Apply K-Means with 3 clusters
final_kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

buyer_data['Cluster'] = final_kmeans.fit_predict(final_features)

print("K-Means Clustering Completed!")

print("\nCluster Distribution:")
display(buyer_data['Cluster'].value_counts().sort_index())
```

```python
# Analyze numerical characteristics of each cluster

cluster_profile = buyer_data.groupby('Cluster').agg(
    Total_Buyers=('client_id', 'count'),
    Avg_Satisfaction=('satisfaction_score', 'mean'),
    Avg_Properties=('properties_count', 'mean'),
    Avg_Sale_Price=('avg_sale_price', 'mean'),
    Avg_Total_Value=('total_sale_value', 'mean'),
    Avg_Max_Price=('max_sale_price', 'mean')
).round(2)

print("Cluster Profile:")
display(cluster_profile)
```

```python
# Acquisition purpose distribution by cluster

purpose_analysis = pd.crosstab(
    buyer_data['Cluster'],
    buyer_data['acquisition_purpose'],
    normalize='index'
).mul(100).round(2)

print("Acquisition Purpose by Cluster (%)")
display(purpose_analysis)

# Client type distribution by cluster

client_analysis = pd.crosstab(
    buyer_data['Cluster'],
    buyer_data['client_type'],
    normalize='index'
).mul(100).round(2)

print("\nClient Type by Cluster (%)")
display(client_analysis)
```

```python
from sklearn.metrics import silhouette_score

# K-Means Silhouette Score
kmeans_score = silhouette_score(
    final_features,
    buyer_data['Cluster'],
    sample_size=1000,
    random_state=42
)

# Hierarchical Silhouette Score
hierarchical_score = silhouette_score(
    final_features,
    buyer_data['Hierarchical_Cluster'],
    sample_size=1000,
    random_state=42
)

print("Clustering Model Comparison")
print("--------------------------------")
print(f"K-Means Silhouette Score: {kmeans_score:.4f}")
print(f"Hierarchical Silhouette Score: {hierarchical_score:.4f}")
```

```python
import json
import os

path = "/content/drive/MyDrive/Colab Notebooks/Project_2_Real_Estate_Buyer_Segmentation.ipynb"

with open(path, "r", encoding="utf-8") as f:
    nb = json.load(f)

for i, cell in enumerate(nb["cells"]):
    source = "".join(cell.get("source", []))

    if "%%writefile app.py" in source:
        print("Dashboard code found in cell:", i)
        source = source.split("\n", 1)[1]

        with open("/content/app.py", "w", encoding="utf-8") as f:
            f.write(source)

        print("App recovered:", os.path.exists("/content/app.py"))
        print("File size:", os.path.getsize("/content/app.py"))
```

```python
import subprocess
import sys
import time
import requests
import os

# Start Streamlit server
log = open("/content/streamlit.log", "w")

process = subprocess.Popen(
    [
        sys.executable, "-m", "streamlit", "run",
        "/content/app.py",
        "--server.port", "8501",
        "--server.address", "0.0.0.0"
    ],
    stdout=log,
    stderr=subprocess.STDOUT,
    start_new_session=True
)

# Wait and check server
for i in range(20):
    time.sleep(2)
    try:
        response = requests.get("http://127.0.0.1:8501", timeout=3)
        print("SUCCESS! Streamlit server running:", response.status_code)
        break
    except:
        print("Starting server...", i + 1)
else:
    print("Server start nahi hua. Log:")
    print(open("/content/streamlit.log").read()[-4000:])
```
